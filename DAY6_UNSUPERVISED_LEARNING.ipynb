```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
print("Libraries Imported Successfully!")
data = {
    "Customer_ID": range(1, 16),
    "Annual_Spending": [
        10000, 12000, 18000, 18000, 20000,
        40000, 45000, 40000, 56000, 59000,
        80000, 87000, 90000, 99000, 100000
    ],
    "Visits_Per_Month": [
        1, 2, 3, 3, 3,
        5, 5, 6, 7, 8,
        10, 11, 12, 12, 13
    ]
}
df = pd.DataFrame(data)
print("\nCustomer Dataset:")
print(df)
X = df[["Annual_Spending", "Visits_Per_Month"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("\nFeatures Scaled Successfully!")
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)
print("\nK-Means Clustering Result:")
print(df)
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Annual_Spending"],
    df["Visits_Per_Month"],
    c=df["Cluster"],
    cmap="viridis",
    s=100
)
plt.xlabel("Annual Spending")
plt.ylabel("Visits Per Month")
plt.title("Customer Segmentation using K-Means")
plt.colorbar(label="Cluster")
plt.show()
average_spending = df.groupby("Cluster")["Annual_Spending"].mean()
print("\nAverage Spending of Each Cluster:")
print(average_spending)
print("\nK-Means Model Comparison:")
for k in [2, 3, 4]:

    model = KMeans(n_clusters=k, random_state=42, n_init=10)

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    print("K =", k)
    print("Inertia (Elbow):", round(model.inertia_, 3))
    print("Silhouette Score:", round(score, 3))
    print()

inertia_values = []

for k in range(1, 7):

    model = KMeans(n_clusters=k, random_state=42, n_init=10)

    model.fit(X_scaled)

    inertia_values.append(model.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(range(1, 7), inertia_values, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
k_values = [2, 3, 4]
silhouette_values = []

for k in k_values:

    model = KMeans(n_clusters=k, random_state=42, n_init=10)

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    silhouette_values.append(score)

plt.figure(figsize=(8, 5))

plt.plot(k_values, silhouette_values, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score Comparison")
plt.show()
dbscan = DBSCAN(eps=0.65, min_samples=3)

df["DBSCAN_Cluster"] = dbscan.fit_predict(X_scaled)

print("\nDBSCAN Clustering Result:")
print(df)
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Annual_Spending"],
    df["Visits_Per_Month"],
    c=df["DBSCAN_Cluster"],
    cmap="plasma",
    s=100
)

plt.xlabel("Annual Spending")
plt.ylabel("Visits Per Month")
plt.title("Customer Segmentation using DBSCAN")
plt.colorbar(label="Cluster")
plt.show()
print("\nFINAL CUSTOMER SEGMENTATION")

print(df)

print("\nPractical 2 Completed Successfully!")
```
